```python

for scene_name in ['bicycle']: #['bicycle','bonsai','counter','flowers','garden','kitchen','treehill','room','stump']:
    
    print(f"scene_name = {scene_name}")
    list_ref_path = glob.glob("/var/home/dotamthuc/Works/Compress3DGS/Results/CompactGS/mip-nerf-360/{}/rate00/test/ours_30000/gt/*.png".format(scene_name))
    scene_sourse = "/var/home/dotamthuc/Works/Compress3DGS/Data/mip-nerf-360/{}/".format(scene_name)
    scene_dataset = Scene(
        scene_sourse, image_dir_name="images",
        res_downscale=0.0, res_width=1600,
        test_view_every=8
    )
    test_views = scene_dataset.get_test_views()

    assert len(list_ref_path) == len(test_views), f"len(list_ref_path)={len(list_ref_path)} == len(test_views)={len(test_views)}"

    num_test_imgs = len(list_ref_path)
    # fig, ax = plt.subplots(nrows=num_test_imgs, ncols=2, figsize=(17, 8*num_test_imgs))
    for i in range(num_test_imgs):
        ref_path = list_ref_path[i]
        img_ref = np.array(Image.open(ref_path)) / 255.0
        img_test = test_views[i].rendered_true.permute(1, 2, 0).numpy()
        
        # ax[i][1].imshow(img_ref)
        # ax[i][0].imshow(img_test)

        assert img_test.shape == img_ref.shape, f"{img_test.shape} != {img_ref.shape}"

        mse = np.mean(np.square(img_test - img_ref))
        psnr = -10 * np.log10(mse+10e-32)

        assert (psnr > 150), "mse = {:.05f} psnr={:.05f}".format(mse, psnr)

        print("mse = {:.05f} psnr={:.05f}".format(mse, psnr))
        # ax[i][1].set_title("mse = {:.05f} psnr={:.05f}".format(mse, psnr))
    # plt.show()
```
